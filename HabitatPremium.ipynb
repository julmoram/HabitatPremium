##**2.Visualización**

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='GrLivArea', y='OverallQual', hue='MSZoning')
plt.title('Relación: Área Habitable vs Calidad (por Zona)')
plt.show()
```

##**3.Sumarización**

```python
# Pivot Table: Promedio de GarageArea por Tipo de Venta y Condición
pivot_casas = df.pivot_table(
    index='SaleType',
    columns='SaleCondition',
    values='GarageArea',
    aggfunc='mean'
).fillna(0) # Rellenamos con 0 donde no haya datos

print("Promedio de Área de Garaje por Categorías:")
display(pivot_casas)
```
